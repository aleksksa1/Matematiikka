```python
#t8 s96
import numpy as np
import matplotlib.pyplot as plt
from sympy.integrals.risch import derivation

x = np.linspace(-1, 1, 100)
y = 0*x+4
x2=np.linspace(1, 3, 100)
y2=-2*x2+6

ax=plt.subplot()

ax.spines['left'].set_position(('data',0))
ax.spines['bottom'].set_position(('data',0))
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')

# samat mittasuhteet
ax.set_aspect('equal')
ax.set_ylim([-1, 5])

plt.plot(x, y, color='blue')
plt.plot(x2,y2, color='blue')

plt.show()
```

```python
#t9
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-5, 0, 100)
y = (3/5)*x+3
x2=np.linspace(0, 4, 100)
y2=0*x2+3
x3 = np.linspace(4, 8, 100)
y3 = -(3/4)*x3+6
x4 = np.linspace(8, 10, 100)
y4 = (1/2)*x4-4

ax=plt.subplot()

ax.spines['left'].set_position(('data',0))
ax.spines['bottom'].set_position(('data',0))
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')

# samat mittasuhteet
ax.set_aspect('equal')
ax.set_ylim([-5, 5])

plt.plot(x, y, color='blue')
plt.plot(x2,y2, color='blue')
plt.plot(x3, y3, color='blue')
plt.plot(x4, y4, color='blue')

plt.show()
```

```python
#s100 t1
#a) ja b)
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-5, 5, 100)
y = x**2-4*x+3
z = -1.5*x**2-3*x+7


ax=plt.subplot()

ax.spines['left'].set_position(('data',0))
ax.spines['bottom'].set_position(('data',0))
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')

# samat mittasuhteet
ax.set_aspect('equal')

# x-akselin säätöä
ax.set_xticks(np.arange(-5, 5, 1))


# Y:lle rajat, että kuvaajat näkyvät fiksummin
plt.ylim([-10, 10])

# label tarvitsee tuon legend():n
plt.plot(x, y, label="a) $y = x^2-4x+3$")
plt.plot(x, z, label="b) $y = -1,5x^2 - 3x+7$")
plt.legend()

plt.show()
```
